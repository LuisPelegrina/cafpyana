```python
import pandas as pd
import pathlib
import glob
import os
import numpy as np
import warnings
import pyanalib.split_df_helpers as splh

# 1. Suppress the annoying warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
from tables import NaturalNameWarning
warnings.filterwarnings('ignore', category=NaturalNameWarning)

def consolidate_dfs_smart_split(folder_path, output_name, max_rows_per_split=500000):
    """
    Consolidates split files and re-splits them into logical chunks 
    to maintain compatibility with splh.load_dfs.
    """
    out_dir = pathlib.Path("/exp/sbnd/data/users/lpelegri/cafpyana_data")
    out_dir.mkdir(parents=True, exist_ok=True)
    
    input_folder = pathlib.Path(folder_path)
    search_pattern = str(input_folder / f"*{output_name}*.df")
    files = sorted(glob.glob(search_pattern))

    if not files:
        print(f"No files found for: {search_pattern}")
        return None

    # 2. DISCOVER KEYS
    with pd.HDFStore(files[0], mode='r') as first_store:
        all_internal_keys = first_store.keys()
    
    keys2load = set()
    for k in all_internal_keys:
        clean_k = k.lstrip('/')
        if clean_k == "split": continue 
        keys2load.add(clean_k.rsplit('_', 1)[0] if '_' in clean_k else clean_k)
    keys2load = sorted(list(keys2load))

    # 3. COLLECT DATA
    master_collector = {key: [] for key in keys2load}
    for file in files:
        print(f"Reading {os.path.basename(file)}...")
        file_data_dict = splh.load_dfs(file, keys2load, n_max_concat=1000)
        for key in keys2load:
            master_collector[key].append(file_data_dict[key])

    # 4. CALCULATE AUTO-SPLIT
    # We find the longest dataframe (usually 'hdr' or 'recTree') 
    # and use it to set n_split for the whole file.
    max_len = 0
    concatenated_data = {}
    for key in keys2load:
        combined = pd.concat(master_collector[key], ignore_index=False)
        concatenated_data[key] = combined
        if len(combined) > max_len:
            max_len = len(combined)

    n_split = int(np.ceil(max_len / max_rows_per_split))
    if n_split == 0: n_split = 1

    # 5. SAVE WITH LEGACY STRUCTURE
    output_file = out_dir / f"{output_name}.df"
    print(f"\nWriting to {output_file} with {n_split} auto-generated splits...")

    with pd.HDFStore(output_file, mode='w') as out_store:
        for key, combined in concatenated_data.items():
            # Break the big dataframe into chunks
            chunks = np.array_split(combined, n_split)
            for i, chunk in enumerate(chunks):
                out_store.put(key=f"{key}_{i}", value=chunk, format="fixed")
            print(f"  - Key '{key}' ({len(combined)} rows) -> saved as _{0}.._{n_split-1}")
        
        # Set n_split metadata for your loading function
        out_store.put(key="split", value=pd.DataFrame({"n_split": [n_split]}), format="fixed")

    print("\nProcess Complete!")
    return output_file

# --- Usage ---
SOURCE = "/exp/sbnd/data/users/lpelegri/cafpyana_data_transfer_folder"
NAME = "cc1pi_5e18_CV_grid_new_cc1pi"

consolidate_dfs_smart_split(SOURCE, NAME)
```
