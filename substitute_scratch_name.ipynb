```python
from __future__ import print_function
import sys
import os
import subprocess
import fnmatch

input_file = "data_lists/AR23p.list"
output_file = "data_lists/AR23p_xroot.list"

substitution = [
    "/pnfs/",
    "root://fndcadoor.fnal.gov:1094/pnfs/fnal.gov/usr/"
]

with open(input_file, 'r') as f_in, open(output_file, 'w') as f_out:
    for line in f_in:
        # Check if line is not just a newline or empty
        if line.strip():
            # Replace the first element of substitution with the second
            new_line = line.replace(substitution[0], substitution[1])
            f_out.write(new_line)

print(f"Done! Remote list saved to {output_file}")
```
