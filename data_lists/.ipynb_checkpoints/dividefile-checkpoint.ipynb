```python
def split_list_file_streaming(input_file, x):
    training_file = "/GiBUU/mc_GIBUU_gen1_first.list"
    final_file    = "mc_MCP2025C_1e20_v10_06_00_09_final.list"

    with open(input_file, "r") as fin, \
         open(training_file, "w") as f_train, \
         open(final_file, "w") as f_final:

        for i, line in enumerate(fin):
            if i < x:
                f_train.write(line)
            else:
                f_final.write(line)
```

```python
split_list_file_streaming("/GiBUU/mc_GIBUU_gen1.list", x=37000)
```

```python
def count_lines(filename):
    with open(filename, "r") as f:
        return sum(1 for _ in f)
```

```python
print(count_lines("mc_MCP2025C_1e20_v10_06_00_09_training.list"))
print(count_lines("mc_MCP2025C_1e20_v10_06_00_09_final.list"))
print(count_lines("mc_MCP2025C_1e20_v10_06_00_09.list"))
```

```python
def find_common_lines(file1, file2, return_lines=False):
    with open(file1, "r") as f1:
        lines1 = set(line.strip() for line in f1 if line.strip())

    common = set()

    with open(file2, "r") as f2:
        for line in f2:
            l = line.strip()
            if l and l in lines1:
                common.add(l)

    if return_lines:
        return common
    else:
        return len(common) > 0
```

```python
find_common_lines("mc_MCP2025C_1e20_v10_06_00_09_training.list", "mc_MCP2025C_1e20_v10_06_00_09_final.list")
```
