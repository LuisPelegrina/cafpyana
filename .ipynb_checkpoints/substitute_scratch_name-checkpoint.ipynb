```python
rom __future__ import print_function
import sys
import os
import subprocess
import fnmatch

Sample = [
    "data_list/Ar23p.list",
    "data_list/"
]

root://fndcadoor.fnal.gov:1094/pnfs/fnal.gov/usr/

outFileList = Sample[0]
inputDir = Sample[1]
file_pattern = "*caf.root"
#file_pattern = "gen_g4_detsim_reco1*Reco2*.root"
#file_pattern = "prodgenie_cosmic_rockbox_sbnd_GenieGen*flat.caf.root"

with open(outFileList, "w") as f:
    # Only look into immediate subdirectories (depth=1)
    for entry in os.listdir(inputDir):
        subdir_path = os.path.join(inputDir, entry)
        if os.path.isdir(subdir_path):
            for file in os.listdir(subdir_path):
                if "root" in file and fnmatch.fnmatch(file, file_pattern):
                    full_path = os.path.join(subdir_path, file)
                    stream = os.popen(f"./my_pnfsToXRootD {full_path}")
                    xroot = stream.read()
                    print(xroot)
                    f.write(xroot)
```
