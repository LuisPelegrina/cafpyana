```python
def apply_selection_and_group(df):
    obvious_cosmic_mask = df.slc.cut.obvious_cosmic
    t0_mask = df.slc.cut.t0
    is_inside_FV_mask = df.slc.cut.inside_FV
    nu_score_mask = df.slc.cut.nu_score
    track_mask = df.slc.cut.track
    shower_mask = df.slc.cut.shower 
    chi2_mask = df.slc.cut.MIP_candidates 
    angle_mask = df.slc.cut.angle 
    proton_BDT_mask = df.slc.cut.proton_BDT
    containment_mask = df.slc.cut.containment 
    michel_mask = df.slc.cut.michel 
    extra_pion_mask = df.slc.cut.extra_pion 
    energy_mask = df.slc.cut.energy
    
    # 1. Define the order of cuts
    cut_sequence = [
        ("cosmic", obvious_cosmic_mask),
        ("t0", t0_mask),
        ("FV", is_inside_FV_mask),
        ("nu_score", nu_score_mask),
        ("track", track_mask),
        ("chi2", chi2_mask),
        ("shower", shower_mask),
        ("angle", angle_mask),
        ("proton_BDT", proton_BDT_mask),
        ("containment", containment_mask),
        ("michel", michel_mask),
        ("extra_pion", extra_pion_mask),
        ("energy", energy_mask)
    ]
    
    # 2. Build the cumulative masks
    cumulative_mak = None
    
    for name, mask in cut_sequence:
        if cumulative_mak is None:
            cumulative_mak = mask
        else:
            cumulative_mak = cumulative_mak & mask

    df = df[cumulative_mak]
    return df.groupby(['__ntuple', 'entry', 'rec.slc..index']).first()
    
```

```python
mc_evt_df = apply_selection_and_group(mc_evt_df)
mc_in_time_cosmics_evt_df = apply_selection_and_group(mc_in_time_cosmics_evt_df)
off_beam_light_evt_df = apply_selection_and_group(off_beam_light_evt_df)
```
